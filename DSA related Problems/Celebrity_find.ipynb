#### Conditions
1. Celebrity is a person who know only him/her.
2. Not know about all persons.
3. All other person know him/her.
4. Celebrity can be only one person.

```python
def celebrity(mat):
        celeb=[]
        for i in range(len(mat)):
            celeb.append(i)

        for i in range(len(mat)):
            for j in range(len(mat[0])):
                if i!=j and mat[i][j]==0:
                    if j in celeb:
                        celeb.remove(j)

        print(celeb)    
        return celeb[0] if celeb and len(celeb)==1 else -1
```

```python
def celebrity(mat):
    n = len(mat)
    for c in range(n):
        is_celeb = True
        for i in range(n):
            if i != c:
                if mat[c][i] != 0 or mat[i][c] != 1:
                    is_celeb = False
                    break
        if is_celeb:
            return c
    return -1
```
