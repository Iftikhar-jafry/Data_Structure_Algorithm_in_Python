get the Kth element from the End in Linkedlist

```python
class Node:
    def __init__(self,value):
        self.data=value
        self.next=None
```

```python
head=Node(1)
N1=Node(2)
N2=Node(3)
N3=Node(4)
N4=Node(5)
```

```python
head.next=N1
N1.next=N2
N2.next=N3
N3.next=N4
N4.next=None
```

```python
def getKthFromLast(head,k):
    my_Node=head
    curr=head
        
    for _ in range(k):
        if curr is None:
            return -1
        curr=curr.next
       
    while curr:
        curr=curr.next
        my_Node=my_Node.next
        
    return my_Node.data
```

```python
getKthFromLast(head,1)
```
