Palindrome in Linkedlist

```python
class Node:
    def __init__(self,value):
        self.data=value
        self.next=None
```

```python
head=Node(1)
N1=Node(2)
N2=Node(3)
N3=Node(4)
N4=Node(5)
```

```python
head.next=N1
N1.next=N2
N2.next=N3
N3.next=N4
N4.next=None
```

```python
def find_palindrom(head):
    li=[]
    while head:
        li.append(head.data)
        head=head.next
    return li==li[: :-1]# for i in range(len(li))-->  { check li[i]!=li[len(li)-1-i] return False } return True
```
