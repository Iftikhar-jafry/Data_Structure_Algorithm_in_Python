```python
def calculateSpan(arr):
    n=len(arr)
    span=[0]*n
    
    for i in range(len(arr)):
        s=0
        j=i
        while j>=0:
            if arr[j]<=arr[i]:
                s+=1
            else:
                break
            j-=1
        span[i]=s
    
    return span
            
```

```python
arr=[100,80,60,70,60,75,85]
```

```python
import time
```

```python
start=time.time()
print(calculateSpan(arr))
end=time.time()
print(end-start)
```

```python
def calculateSpan(price):
    n = len(price)
    span = [0] * n
    stack = []  # will store indices

    for i in range(n):
        # Pop prices from stack while current price is higher
        while stack and price[i] >= price[stack[-1]]:
            stack.pop()

        # If stack is empty, all previous prices are smaller
        span[i] = i + 1 if not stack else i - stack[-1]

        # Push current index to stack
        stack.append(i)

    return span
```
