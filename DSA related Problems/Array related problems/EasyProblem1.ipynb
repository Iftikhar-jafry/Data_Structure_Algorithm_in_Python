Indexes of Subarray Sum


```python
arr=[2,4,5,1,3,7]
```

```python
def Target_Sum(arr,target):
    for i in range(len(arr)):
        current_sum=0
        for j in range(i,len(arr)):
            current_sum+=arr[j]
            if current_sum==target:
                return [i,j]
            if current_sum>target: # To Not check sum becomes greater than target
                break
    
    return f"No Subarray with sum {target}"
                
```

```python
Target_Sum(arr,20)
```
