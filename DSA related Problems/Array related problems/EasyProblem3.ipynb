```python
def getSecondLargest1(arr):
        if len(arr)>1:
            arr=sorted(list(set(arr)))
            return arr[len(arr)-2]
        return -1
```

```python
getSecondLargest1([2,3,8,7,3,1,8])
```

```python
def getSecondLargest2(arr):
        if len(arr)==1:
            return -1
        sec_val=fir_val=float('-inf')
        for num in arr:
            if fir_val<num:
                sec_val=fir_val
                fir_val=num
            elif fir_val>num>sec_val:
                 sec_val=num
        return sec_val if sec_val!=float('-inf') else -1
```
