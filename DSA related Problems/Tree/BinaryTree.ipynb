#### Binary Tree

```python
class Node:
    def __init__(self,value):
        self.data=value
        self.left=None
        self.right=None
```

```python
class Binary_Tree:
    def __init__(self):
        self.root=None
        self.n=0
    
    def insert_value(self,value):
        if self.root is None:
            self.root=Node(value)
            self.n+=1
            return
        self.insert(self.root,value)

    def insert(self,curr,value):
        if value<curr.data:
            if curr.left is None:
                curr.left=Node(value)
                self.n+=1
            else:
                self.insert(curr.left,value)
        else:
            if curr.right is None:
                curr.right=Node(value)
                self.n+=1
            else:
                self.insert(curr.right,value)
    def Pre_Order(self):
        if self.root is None:
            return "Empty Tree"
        self.Pre_Order_traversal(self.root)
    
    def Pre_Order_traversal(self,curr):
        if curr is None:
            return 
        print(curr.data , end=" ")

        self.Pre_Order_traversal(curr.left)
        self.Pre_Order_traversal(curr.right)
    
    def Post_Order(self):
        if self.root is None:
            return "Empty Tree"
        self.Post_Order_traversal(self.root)
    
    def Post_Order_traversal(self,curr):
        if curr is None:
            return 
        self.Post_Order_traversal(curr.left)
        self.Post_Order_traversal(curr.right)
        print(curr.data , end=" ")
    
    def In_Order(self):
        if self.root is None:
            return "Empty Tree"
        self.In_Order_traversal(self.root)
    
    def In_Order_traversal(self,curr):
        if curr is None:
            return 
        self.In_Order_traversal(curr.left)
        print(curr.data , end=" ")
        self.In_Order_traversal(curr.right)
    
    def Level_Order(self):
        if self.root is None:
            return 'Empty Tree'
        Queue=[]
        Queue.append(self.root)
        while len(Queue)>0:
            node=Queue.pop(0)
            print(node.data, end=" ")

            if node.left is not None:
                Queue.append(node.left)
            if node.right is not None:
                Queue.append(node.right) 
```

```python
BT=Binary_Tree()
```

```python
BT.insert_value(8)
BT.insert_value(6)
BT.insert_value(7)
BT.insert_value(5)
BT.insert_value(10)
BT.insert_value(9)
BT.insert_value(11)
```

```python
BT.Pre_Order()
print('\n')
BT.Post_Order()
print('\n')
BT.In_Order()
print('\n')
BT.Level_Order()
```
