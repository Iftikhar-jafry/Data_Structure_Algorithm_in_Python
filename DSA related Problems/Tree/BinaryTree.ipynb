#### Binary Tree
8 <code> root </code> <br>
|-->6 <code> left child of 8 </code><br>
|&nbsp;&nbsp;&nbsp; |-->5 <code> left child of 6 </code><br>
|&nbsp;&nbsp;&nbsp; |-->7 <code> right child of 6 </code><br>
|<br>
|-->10 <code> right child of 8 </code><br>
&nbsp;&nbsp;&nbsp;&nbsp; |-->9 <code> left child of 10 </code><br>
&nbsp;&nbsp;&nbsp;&nbsp; |-->11 <code> right child of 10 </code><br>

```python
class Node:
    def __init__(self,value):
        self.data=value
        self.left=None
        self.right=None
```

```python
class Binary_Tree:
    def __init__(self):
        self.root=None
        self.n=0
    
    def insert_value(self,value):
        if self.root is None:
            self.root=Node(value)
            self.n+=1
            return
        self.insert(self.root,value)

    def insert(self,curr,value):
        if value<curr.data:
            if curr.left is None:
                curr.left=Node(value)
                self.n+=1
            else:
                self.insert(curr.left,value)
        else:
            if curr.right is None:
                curr.right=Node(value)
                self.n+=1
            else:
                self.insert(curr.right,value)
    def Pre_Order(self):
        if self.root is None:
            return "Empty Tree"
        self.Pre_Order_traversal(self.root)
    
    def Pre_Order_traversal(self,curr):
        if curr is None:
            return 
        print(curr.data , end=" ")

        self.Pre_Order_traversal(curr.left)
        self.Pre_Order_traversal(curr.right)
    
    def Post_Order(self):
        if self.root is None:
            return "Empty Tree"
        self.Post_Order_traversal(self.root)
    
    def Post_Order_traversal(self,curr):
        if curr is None:
            return 
        self.Post_Order_traversal(curr.left)
        self.Post_Order_traversal(curr.right)
        print(curr.data , end=" ")
    
    def In_Order(self):
        if self.root is None:
            return "Empty Tree"
        self.In_Order_traversal(self.root)
    
    def In_Order_traversal(self,curr):
        if curr is None:
            return 
        self.In_Order_traversal(curr.left)
        print(curr.data , end=" ")
        self.In_Order_traversal(curr.right)
    
    def Level_Order(self):
        if self.root is None:
            return 'Empty Tree'
        Queue=[]
        Queue.append(self.root)
        while len(Queue)>0:
            node=Queue.pop(0)
            print(node.data, end=" ")

            if node.left is not None:
                Queue.append(node.left)
            if node.right is not None:
                Queue.append(node.right)
    def Height(self):
        if self.root is None:
            return 0
        return self.Height_of_tree(self.root)
    def Height_of_tree(self,curr):
        if curr is None:
            return 0
        left_height=self.Height_of_tree(curr.left)
        right_height=self.Height_of_tree(curr.right)
        return max(left_height,right_height)+1
    def Size(self):
        if self.root is None:
            return 0
        return self.Size_of_tree(self.root)
    def Size_of_tree(self,curr):
        if curr is None:
            return 0
        
        left_size=self.Size_of_tree(curr.left)
        right_size=self.Size_of_tree(curr.right)
        return left_size+right_size+1
    
    
    def Search(self,value):
        if self.root is None:
            return "Empty Tree"
        
        return self.Search_Node(self.root,value)
    def Search_Node(self,curr,value):
        if curr is None:
            return False
        if curr.data == value:
            return True

        if curr.data > value:
            return self.Search_Node(curr.left,value)
        else:
            return self.Search_Node(curr.right,value)
       
    def Delete(self,value):
        if self.root is None:
            return False
        self.root=self.Delete_from_tree(self.root,value)
        return True
    def Delete_from_tree(self,curr,value):
        if curr is None:
            return curr
        if value<curr.data:
            curr.left=self.Delete_from_tree(curr.left,value)
        elif value>curr.data:
            curr.right=self.Delete_from_tree(curr.right,value)
        else:
            if curr.left is None:
                return curr.right
            elif curr.right is None:
                return curr.left
            temp=self.Min_value_node(curr.right)
            curr.data=temp.data
            curr.right=self.Delete_from_tree(curr.right,temp.data)
        return curr
    def Min_value_node(self,curr):
        if curr is None or curr.left is None:
            return curr
        return self.Min_value_node(curr.left)
    
    def find_min_value(self):
        if self.root is None:
            return 'Empty Tree'
        curr=self.root
        while curr.left is not None:
            curr=curr.left
        return curr.data
    def find_max_value(self):
        if self.root is None:
            return 'Empty Tree'
        curr=self.root
        while curr.right is not None:
            curr=curr.right
        return curr.data
    
    def is_leaf_node(self,value):
        if self.root is None:
            return 'Empty Tree'
        
        return self.leaf_node(self.root,value)
    
    def leaf_node(self,curr,value):
        if curr is None:
            return False
        if curr.data== value:
            if curr.left is None and curr.right is None:
                return True
        if curr.data>value:
            return self.leaf_node(curr.left,value)
        else:
            return self.leaf_node(curr.right,value)
        
    def is_Ancester_node(self,ances,value):
        if self.root is None:
            return False
        
        return self.Ancester_Node(self.root, ances,value)
    def Ancester_Node(self,curr,ances,value):
        if curr is None:
            return False
        
        if curr.data==ances:
            return self.Search_Node(curr,value)
        if curr.data>ances:
            return self.Ancester_Node(curr.left,ances,value)
        else:
            return self.Ancester_Node(curr.right,ances,value)
    
    def Mirror(self):
        if self.root is None:
            return 'Empty Tree'
        return self.Mirror_Tree(self.root.left , self.root.right)
    def Mirror_Tree(self,l,r):
        if l is None and r is None:
            return True
        if l is not None and r is not None:
            if l.data==r.data:
                return True
            else:
                return False
        else:
            return False
        
        half1= self.Mirror_Tree(l.left,r.right)
        half2=self.Mirror_Tree(l.right,r.left)

        return half1==half2
```
